```python
# Install required packages
!pip install -q transformers datasets torch scikit-learn pandas accelerate scipy seaborn matplotlib
print("📦 Packages installed!")

# Mount and navigate
from google.colab import drive
drive.mount('/content/drive')
import os
os.chdir('/content/drive/MyDrive/yakut-sa')
print(f"📂 Working directory: {os.getcwd()}")

import torch
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.metrics import accuracy_score, classification_report, f1_score, confusion_matrix
from sklearn.utils.class_weight import compute_class_weight
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    EarlyStoppingCallback,
    DataCollatorWithPadding
)
from datasets import Dataset
import json
import warnings
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings('ignore')

# ----------------------------
# Reproducibility
# ----------------------------
def set_seed(seed: int = 42):
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

# ----------------------------
# Config (adapted for bidirectional cross-dataset)
# ----------------------------
CONFIG = {
    'model_name': "bert-base-multilingual-cased",
    'use_extended_tokenizer': False,  # Set to True to use extended tokenizer
    'custom_tokenizer_path': "models/mbert_extended_wp_sakha",

    # Dataset files
    'dataset_a': "data/annotated/kyym_ann.csv",
    'dataset_b': "data/annotated/forum_ykt_ann.csv",
    'outdir': "results_bidirectional_cross_dataset",

    # Training parameters
    'epochs': 5,
    'batch_size': 8,
    'lr': 1e-5,
    'max_length': 210,
    'seed': 42,

    # Class balancing
    'use_class_weights': True,

    # Early stopping
    'early_stopping_patience': 2,
    'early_stopping_threshold': 0.001,

    # Training parameters
    'label_smoothing': 0.05,
    'warmup_ratio': 0.06,
    'weight_decay': 0.01,

    # Memory/throughput safety
    'gradient_accumulation_steps': 2,
    'fp16': True,

    # Logging/eval/save by steps
    'logging_steps': 20,
    'eval_strategy': "steps",
    'eval_steps': 40,
    'save_strategy': "steps",
    'save_steps': 40,

    'metric_for_best_model': "eval_f1",
    'greater_is_better': True,
    'load_best_model_at_end': True,
    'save_total_limit': 2,
}
set_seed(CONFIG['seed'])

# ----------------------------
# Dataset analysis and verification
# ----------------------------
def _label_col(df):
    """Find the label column in dataframe"""
    cols = [c.lower().strip() for c in df.columns]
    if "label" in cols:
        return "label"
    if "sentiment" in cols:
        return "sentiment"
    raise ValueError("No 'label' or 'sentiment' column in dataframe.")

def normalize_label(label):
    """Normalize labels by converting to lowercase and stripping whitespace"""
    if pd.isna(label):
        return label
    return str(label).lower().strip()

def analyze_dataset_pair(train_path, test_path, direction_name):
    """Analyze a pair of datasets for compatibility"""
    print(f"\n{'='*60}")
    print(f"ANALYZING {direction_name}")
    print(f"{'='*60}")
    print(f"Training on: {train_path}")
    print(f"Testing on:  {test_path}")

    # Load datasets
    if not Path(train_path).exists():
        raise FileNotFoundError(f"Training dataset not found: {train_path}")
    if not Path(test_path).exists():
        raise FileNotFoundError(f"Test dataset not found: {test_path}")

    train_df = pd.read_csv(train_path)
    test_df = pd.read_csv(test_path)

    train_label_col = _label_col(train_df)
    test_label_col = _label_col(test_df)

    print(f"\n📈 Training Dataset ({train_path}):")
    print(f"  - Size: {len(train_df):,} samples")
    print(f"  - Label column: '{train_label_col}'")
    print(f"  - Raw labels: {sorted(train_df[train_label_col].unique())}")
    print(f"  - Label distribution: {dict(train_df[train_label_col].value_counts())}")
    print(f"  - Avg text length: {train_df['text'].str.len().mean():.1f} chars")

    print(f"\n📉 Test Dataset ({test_path}):")
    print(f"  - Size: {len(test_df):,} samples")
    print(f"  - Label column: '{test_label_col}'")
    print(f"  - Raw labels: {sorted(test_df[test_label_col].unique())}")
    print(f"  - Label distribution: {dict(test_df[test_label_col].value_counts())}")
    print(f"  - Avg text length: {test_df['text'].str.len().mean():.1f} chars")

    # Normalize labels for compatibility checking
    train_labels_norm = set(normalize_label(label) for label in train_df[train_label_col].unique())
    test_labels_norm = set(normalize_label(label) for label in test_df[test_label_col].unique())

    # Remove any NaN labels
    train_labels_norm.discard(None)
    test_labels_norm.discard(None)

    print(f"\n🔄 After normalization (lowercase, stripped):")
    print(f"  - Training labels: {sorted(train_labels_norm)}")
    print(f"  - Test labels: {sorted(test_labels_norm)}")

    # Check compatibility after normalization
    if train_labels_norm != test_labels_norm:
        print(f"\n⚠️  LABEL MISMATCH DETECTED (after normalization):")
        print(f"  - Only in training: {train_labels_norm - test_labels_norm}")
        print(f"  - Only in test: {test_labels_norm - train_labels_norm}")
        print(f"  - Common labels: {train_labels_norm & test_labels_norm}")

        if not (train_labels_norm & test_labels_norm):
            raise ValueError("No common labels between datasets even after normalization!")
        else:
            print("⚠️  Proceeding with common labels only - some data will be lost!")
    else:
        print(f"\n✅ Label sets are compatible after normalization")

    # Show sample texts
    print(f"\n📝 Sample texts:")
    print(f"Training sample: '{train_df.iloc[0]['text'][:100]}...'")
    print(f"Test sample: '{test_df.iloc[0]['text'][:100]}...'")

    return train_df, test_df, train_label_col, test_label_col

def load_and_prepare_dataset_pair(train_path, test_path):
    """Load and prepare a pair of datasets with unified label encoding"""
    train_df, test_df, train_label_col, test_label_col = analyze_dataset_pair(
        train_path, test_path, f"Direction: {Path(train_path).stem} → {Path(test_path).stem}"
    )

    # Normalize labels in both datasets
    train_df = train_df.copy()
    test_df = test_df.copy()
    train_df[train_label_col + '_normalized'] = train_df[train_label_col].apply(normalize_label)
    test_df[test_label_col + '_normalized'] = test_df[test_label_col].apply(normalize_label)

    # Create unified label encoding from normalized labels
    train_labels_norm = set(train_df[train_label_col + '_normalized'].dropna().unique())
    test_labels_norm = set(test_df[test_label_col + '_normalized'].dropna().unique())
    all_labels = sorted(train_labels_norm | test_labels_norm)

    label_encoder = {label: idx for idx, label in enumerate(all_labels)}
    reverse_encoder = {idx: label for label, idx in label_encoder.items()}

    print(f"\n🔢 Unified label encoding: {label_encoder}")

    # Clean and prepare training data using normalized labels
    train_clean = train_df[['text', train_label_col + '_normalized']].dropna().copy()
    train_clean = train_clean.rename(columns={train_label_col + '_normalized': 'label_text'})
    train_clean['label'] = train_clean['label_text'].map(label_encoder)

    # Clean and prepare test data using normalized labels
    test_clean = test_df[['text', test_label_col + '_normalized']].dropna().copy()
    test_clean = test_clean.rename(columns={test_label_col + '_normalized': 'label_text'})
    test_clean['label'] = test_clean['label_text'].map(label_encoder)

    # Check for unmapped labels
    if train_clean['label'].isna().any():
        bad_labels = train_clean[train_clean['label'].isna()]['label_text'].unique()
        raise ValueError(f"Unmapped training labels: {bad_labels}")

    if test_clean['label'].isna().any():
        bad_labels = test_clean[test_clean['label'].isna()]['label_text'].unique()
        raise ValueError(f"Unmapped test labels: {bad_labels}")

    print(f"\n📊 Final dataset sizes after cleaning:")
    print(f"  Training: {len(train_clean):,} samples")
    print(f"  Test:     {len(test_clean):,} samples")

    # Show final label distributions
    print(f"\n📈 Final label distributions:")
    print("Training:")
    for label in sorted(label_encoder.keys()):
        count = int((train_clean['label'] == label_encoder[label]).sum())
        print(f"  {label}: {count}")

    print("Test:")
    for label in sorted(label_encoder.keys()):
        count = int((test_clean['label'] == label_encoder[label]).sum())
        print(f"  {label}: {count}")

    # Compute class weights from training data only
    class_weights = None
    if CONFIG['use_class_weights']:
        labels = train_clean['label'].astype(int).values
        classes = np.unique(labels)
        weights = compute_class_weight('balanced', classes=classes, y=labels)
        class_weights = {int(c): float(w) for c, w in zip(classes, weights)}
        print(f"📊 Class weights (from training data): {class_weights}")

    return (train_clean[['text', 'label']],
            test_clean[['text', 'label']],
            label_encoder,
            reverse_encoder,
            class_weights)

# ----------------------------
# Tokenization
# ----------------------------
def tokenize_dataset(df: pd.DataFrame, tokenizer, max_length: int):
    """Tokenize dataset with dynamic padding"""
    def tokenize_function(examples):
        return tokenizer(
            examples['text'],
            truncation=True,
            max_length=max_length,
            return_tensors=None
        )

    df_copy = df.copy()
    df_copy['label'] = df_copy['label'].astype(int)
    dataset = Dataset.from_pandas(df_copy[['text', 'label']], preserve_index=False)
    tokenized = dataset.map(
        tokenize_function,
        batched=True,
        remove_columns=['text'],
        desc="Tokenizing"
    )
    return tokenized

# ----------------------------
# Weighted Trainer
# ----------------------------
class WeightedTrainer(Trainer):
    def __init__(self, *args, class_weights=None, **kwargs):
        super().__init__(*args, **kwargs)
        self.class_weights = class_weights

    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None):
        labels = inputs.pop("labels")
        outputs = model(**inputs)
        logits = outputs.logits if hasattr(outputs, "logits") else outputs[0]

        if self.class_weights is not None:
            num_labels = model.config.num_labels
            weight_vec = torch.ones(num_labels, dtype=torch.float32, device=logits.device)
            for cls_id, w in self.class_weights.items():
                if 0 <= int(cls_id) < num_labels:
                    weight_vec[int(cls_id)] = float(w)

            loss_fct = torch.nn.CrossEntropyLoss(
                weight=weight_vec,
                label_smoothing=self.args.label_smoothing_factor if hasattr(self.args, "label_smoothing_factor") else 0.0,
            )
        else:
            loss_fct = torch.nn.CrossEntropyLoss(
                label_smoothing=self.args.label_smoothing_factor if hasattr(self.args, "label_smoothing_factor") else 0.0,
            )

        loss = loss_fct(logits.view(-1, model.config.num_labels), labels.view(-1))
        return (loss, outputs) if return_outputs else loss

# ----------------------------
# Single direction training
# ----------------------------
def train_single_direction(train_path, test_path, direction_name, tokenizer):
    """Train in one direction and return results"""
    print(f"\n{'='*80}")
    print(f"TRAINING {direction_name}")
    print(f"{'='*80}")

    # Load and prepare data
    train_df, test_df, label_encoder, reverse_encoder, class_weights = load_and_prepare_dataset_pair(
        train_path, test_path
    )
    num_labels = len(label_encoder)

    # Load model
    print(f"\n🤖 Loading fresh model...")
    model = AutoModelForSequenceClassification.from_pretrained(
        CONFIG['model_name'],
        num_labels=num_labels,
        ignore_mismatched_sizes=True
    )

    # Resize embeddings if needed
    if len(tokenizer) != model.config.vocab_size:
        print(f"🔧 Resizing embeddings: {model.config.vocab_size} → {len(tokenizer)}")
        model.resize_token_embeddings(len(tokenizer))

    # Tokenize datasets
    print(f"\n🔄 Tokenizing datasets...")
    train_dataset = tokenize_dataset(train_df, tokenizer, CONFIG['max_length'])
    test_dataset = tokenize_dataset(test_df, tokenizer, CONFIG['max_length'])

    # Create validation split from training data (20%)
    train_size = int(0.8 * len(train_dataset))
    val_size = len(train_dataset) - train_size
    train_dataset, val_dataset = torch.utils.data.random_split(
        train_dataset, [train_size, val_size],
        generator=torch.Generator().manual_seed(CONFIG['seed'])
    )

    print(f"📊 Dataset splits: {len(train_dataset)} train, {len(val_dataset)} validation, {len(test_dataset)} test")

    # Training arguments
    output_dir = Path(CONFIG['outdir']) / direction_name.lower().replace(' ', '_').replace('→', '_to_')
    output_dir.mkdir(exist_ok=True, parents=True)

    training_args = TrainingArguments(
        output_dir=str(output_dir),
        num_train_epochs=CONFIG['epochs'],
        per_device_train_batch_size=CONFIG['batch_size'],
        per_device_eval_batch_size=max(CONFIG['batch_size'], 16),
        gradient_accumulation_steps=CONFIG['gradient_accumulation_steps'],

        learning_rate=CONFIG['lr'],
        warmup_ratio=CONFIG['warmup_ratio'],
        weight_decay=CONFIG['weight_decay'],

        eval_strategy=CONFIG['eval_strategy'],
        eval_steps=CONFIG['eval_steps'],
        save_strategy=CONFIG['save_strategy'],
        save_steps=CONFIG['save_steps'],

        logging_steps=CONFIG['logging_steps'],
        save_total_limit=CONFIG['save_total_limit'],
        load_best_model_at_end=CONFIG['load_best_model_at_end'],
        metric_for_best_model=CONFIG['metric_for_best_model'],
        greater_is_better=CONFIG['greater_is_better'],

        fp16=CONFIG['fp16'] and torch.cuda.is_available(),
        label_smoothing_factor=CONFIG['label_smoothing'],

        report_to=[],
        seed=CONFIG['seed'],
        push_to_hub=False,

        gradient_checkpointing=True,
        # FIX: Only enable TF32 on Ampere+ GPUs (compute capability >= 8.0)
        tf32=torch.cuda.is_available() and torch.cuda.get_device_capability()[0] >= 8,
    )

    # Metrics
    def compute_metrics(eval_pred):
        if isinstance(eval_pred, tuple):
            preds, labels = eval_pred
        else:
            preds, labels = eval_pred.predictions, eval_pred.label_ids
        if isinstance(preds, (list, tuple)):
            preds = preds[0]
        predictions = np.argmax(preds, axis=1)

        accuracy = accuracy_score(labels, predictions)
        f1_macro = f1_score(labels, predictions, average='macro')
        f1_weighted = f1_score(labels, predictions, average='weighted')
        return {"accuracy": accuracy, "f1": f1_weighted, "f1_macro": f1_macro}

    # Create trainer
    trainer = WeightedTrainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=val_dataset,
        tokenizer=tokenizer,
        data_collator=DataCollatorWithPadding(tokenizer=tokenizer),
        compute_metrics=compute_metrics,
        class_weights=class_weights,
        callbacks=[EarlyStoppingCallback(
            early_stopping_patience=CONFIG['early_stopping_patience'],
            early_stopping_threshold=CONFIG['early_stopping_threshold']
        )]
    )

    # Train
    print(f"\n🚀 Starting training...")
    train_result = trainer.train()

    # Evaluate on validation set
    print(f"\n📊 Evaluating on validation set...")
    val_result = trainer.evaluate()

    # Test on cross-dataset
    print(f"\n🎯 Testing on cross-dataset...")
    test_predictions = trainer.predict(test_dataset)
    y_pred = np.argmax(test_predictions.predictions, axis=1)
    y_true = test_predictions.label_ids

    # Calculate cross-dataset metrics
    test_accuracy = accuracy_score(y_true, y_pred)
    test_f1_weighted = f1_score(y_true, y_pred, average='weighted')
    test_f1_macro = f1_score(y_true, y_pred, average='macro')

    # Detailed classification report
    target_names = [reverse_encoder[i] for i in sorted(reverse_encoder.keys())]
    print(f"\n📈 Cross-Dataset Classification Report ({direction_name}):")
    print(classification_report(y_true, y_pred, target_names=target_names, digits=4))

    # Confusion matrix
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=target_names, yticklabels=target_names)
    plt.title(f'Cross-Dataset Confusion Matrix ({direction_name})')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.savefig(output_dir / 'confusion_matrix.png', dpi=300, bbox_inches='tight')
    plt.show()

    # Save best model
    if CONFIG['load_best_model_at_end']:
        trainer.save_model(str(output_dir / "best_model"))
        tokenizer.save_pretrained(str(output_dir / "best_model"))

    # Calculate performance drop
    val_acc = val_result['eval_accuracy']
    val_f1 = val_result['eval_f1']
    acc_drop = (val_acc - test_accuracy) / val_acc * 100 if val_acc > 0 else 0
    f1_drop = (val_f1 - test_f1_weighted) / val_f1 * 100 if val_f1 > 0 else 0

    print(f"\n📊 {direction_name} Results Summary:")
    print(f"  Validation Performance:  Acc={val_acc:.4f}, F1={val_f1:.4f}")
    print(f"  Cross-Dataset Performance: Acc={test_accuracy:.4f}, F1={test_f1_weighted:.4f}")
    print(f"  Performance Drop:        Acc={acc_drop:.1f}%, F1={f1_drop:.1f}%")

    # Cleanup
    del model, trainer
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return {
        'direction': direction_name,
        'train_dataset': train_path,
        'test_dataset': test_path,
        'train_samples': len(train_df),
        'test_samples': len(test_df),
        'num_labels': num_labels,
        'label_encoder': label_encoder,

        'validation_metrics': {
            'accuracy': float(val_result['eval_accuracy']),
            'f1_weighted': float(val_result['eval_f1']),
            'f1_macro': float(val_result['eval_f1_macro'])
        },

        'cross_dataset_metrics': {
            'accuracy': float(test_accuracy),
            'f1_weighted': float(test_f1_weighted),
            'f1_macro': float(test_f1_macro)
        },

        'performance_drops': {
            'accuracy_drop_percent': float(acc_drop),
            'f1_drop_percent': float(f1_drop)
        },

        'predictions': y_pred.tolist(),
        'true_labels': y_true.tolist()
    }

# ----------------------------
# Bidirectional experiment
# ----------------------------
def run_bidirectional_experiment():
    """Run bidirectional cross-dataset experiment"""
    print("🚀 Starting Bidirectional Cross-Dataset Yakut Sentiment Analysis")
    print("="*80)

    # Load tokenizer
    print(f"\n🔧 Loading tokenizer...")
    tokenizer_path = CONFIG['custom_tokenizer_path']
    if Path(tokenizer_path).exists():
        tokenizer = AutoTokenizer.from_pretrained(tokenizer_path, use_fast=True)
        print(f"✅ Custom tokenizer loaded: {len(tokenizer):,} tokens")
    else:
        tokenizer = AutoTokenizer.from_pretrained(CONFIG['model_name'], use_fast=True)
        print(f"✅ Default tokenizer loaded: {len(tokenizer):,} tokens")

    # Define both directions
    dataset_a = CONFIG['dataset_a']  # kyym_ann.csv
    dataset_b = CONFIG['dataset_b']  # forum_ykt_ann.csv

    directions = [
        {
            'name': f"{Path(dataset_a).stem} → {Path(dataset_b).stem}",
            'train': dataset_a,
            'test': dataset_b,
            'desc': f"Train on {Path(dataset_a).stem}, test on {Path(dataset_b).stem}"
        },
        {
            'name': f"{Path(dataset_b).stem} → {Path(dataset_a).stem}",
            'train': dataset_b,
            'test': dataset_a,
            'desc': f"Train on {Path(dataset_b).stem}, test on {Path(dataset_a).stem}"
        }
    ]

    print(f"\n🔄 Will run {len(directions)} directions:")
    for i, direction in enumerate(directions, 1):
        print(f"  {i}. {direction['name']}: {direction['desc']}")

    # Run both directions
    all_results = []
    for direction in directions:
        try:
            print(f"\n" + "="*20 + f" STARTING {direction['name']} " + "="*20)
            result = train_single_direction(
                direction['train'],
                direction['test'],
                direction['name'],
                tokenizer
            )
            all_results.append(result)
            print(f"✅ {direction['name']} completed successfully!")

        except Exception as e:
            print(f"❌ {direction['name']} failed: {e}")
            import traceback
            traceback.print_exc()

    # Comparative analysis
    if len(all_results) == 2:
        print(f"\n{'='*80}")
        print("BIDIRECTIONAL COMPARATIVE ANALYSIS")
        print(f"{'='*80}")

        result_a, result_b = all_results

        print(f"\n📊 Direction Comparison:")
        print(f"  {result_a['direction']}:")
        print(f"    Cross-dataset Accuracy:  {result_a['cross_dataset_metrics']['accuracy']:.4f}")
        print(f"    Cross-dataset F1:        {result_a['cross_dataset_metrics']['f1_weighted']:.4f}")
        print(f"    Performance Drop (Acc):  {result_a['performance_drops']['accuracy_drop_percent']:.1f}%")
        print(f"    Performance Drop (F1):   {result_a['performance_drops']['f1_drop_percent']:.1f}%")

        print(f"\n  {result_b['direction']}:")
        print(f"    Cross-dataset Accuracy:  {result_b['cross_dataset_metrics']['accuracy']:.4f}")
        print(f"    Cross-dataset F1:        {result_b['cross_dataset_metrics']['f1_weighted']:.4f}")
        print(f"    Performance Drop (Acc):  {result_b['performance_drops']['accuracy_drop_percent']:.1f}%")
        print(f"    Performance Drop (F1):   {result_b['performance_drops']['f1_drop_percent']:.1f}%")

        # Asymmetry analysis
        acc_diff = result_a['cross_dataset_metrics']['accuracy'] - result_b['cross_dataset_metrics']['accuracy']
        f1_diff = result_a['cross_dataset_metrics']['f1_weighted'] - result_b['cross_dataset_metrics']['f1_weighted']
        drop_diff_acc = result_b['performance_drops']['accuracy_drop_percent'] - result_a['performance_drops']['accuracy_drop_percent']
        drop_diff_f1 = result_b['performance_drops']['f1_drop_percent'] - result_a['performance_drops']['f1_drop_percent']

        print(f"\n🔄 Transfer Asymmetry Analysis:")
        print(f"  Accuracy difference:     {acc_diff:+.4f} (A - B)")
        print(f"  F1 difference:           {f1_diff:+.4f} (A - B)")
        print(f"  Drop difference (Acc):   {drop_diff_acc:+.1f}% (B - A)")
        print(f"  Drop difference (F1):    {drop_diff_f1:+.1f}% (B - A)")

        # Interpretations
        print(f"\n🧠 Domain Transfer Insights:")

        if abs(acc_diff) < 0.05:
            print("  • Symmetric transfer performance - datasets are similarly complex")
        elif acc_diff > 0:
            print(f"  • Direction A performs better by {acc_diff:.3f}")
            print(f"    → {Path(result_a['train_dataset']).stem} may be more generalizable as training data")
            print(f"    → {Path(result_a['test_dataset']).stem} may be an easier target domain")
        else:
            print(f"  • Direction B performs better by {abs(acc_diff):.3f}")
            print(f"    → {Path(result_b['train_dataset']).stem} may be more generalizable as training data")
            print(f"    → {Path(result_b['test_dataset']).stem} may be an easier target domain")

        if abs(drop_diff_acc) < 5:
            print("  • Similar generalization difficulty in both directions")
        elif drop_diff_acc > 0:
            print(f"  • Direction B shows {drop_diff_acc:.1f}% more performance drop")
            print(f"    → Transferring TO {Path(result_b['test_dataset']).stem} is harder")
        else:
            print(f"  • Direction A shows {abs(drop_diff_acc):.1f}% more performance drop")
            print(f"    → Transferring TO {Path(result_a['test_dataset']).stem} is harder")

        # Quality assessment
        min_cross_acc = min(result_a['cross_dataset_metrics']['accuracy'],
                           result_b['cross_dataset_metrics']['accuracy'])
        max_drop = max(result_a['performance_drops']['accuracy_drop_percent'],
                      result_b['performance_drops']['accuracy_drop_percent'])

        print(f"\n🎯 Overall Assessment:")
        if min_cross_acc > 0.7:
            print("  • Strong cross-dataset generalization achieved")
        elif min_cross_acc > 0.5:
            print("  • Moderate cross-dataset generalization achieved")
        else:
            print("  • Poor cross-dataset generalization - significant domain shift detected")

        if max_drop < 10:
            print("  • Low generalization gap - datasets are well-aligned")
        elif max_drop < 25:
            print("  • Moderate generalization gap - some domain adaptation needed")
        else:
            print("  • High generalization gap - significant domain differences")

    # Save comprehensive results
    output_dir = Path(CONFIG['outdir'])
    output_dir.mkdir(exist_ok=True, parents=True)

    final_results = {
        'config': CONFIG,
        'experiment_type': 'bidirectional_cross_dataset',
        'dataset_a': CONFIG['dataset_a'],
        'dataset_b': CONFIG['dataset_b'],
        'results': all_results,
        'summary': {
            'num_directions': len(all_results),
            'successful_directions': len(all_results)
        }
    }

    if len(all_results) == 2:
        final_results['comparative_analysis'] = {
            'accuracy_difference': float(acc_diff),
            'f1_difference': float(f1_diff),
            'drop_difference_accuracy': float(drop_diff_acc),
            'drop_difference_f1': float(drop_diff_f1),
            'better_direction': result_a['direction'] if acc_diff > 0 else result_b['direction'],
            'min_cross_accuracy': float(min_cross_acc),
            'max_performance_drop': float(max_drop)
        }

    results_file = output_dir / 'bidirectional_results.json'
    with open(results_file, 'w') as f:
        json.dump(final_results, f, indent=2)

    print(f"\n💾 Complete results saved to {results_file}")

    return final_results

# ----------------------------
# Main
# ----------------------------
def main():
    try:
        results = run_bidirectional_experiment()
        print("\n✅ Bidirectional cross-dataset experiment completed successfully!")
        return results
    except Exception as e:
        print(f"❌ Experiment failed: {e}")
        import traceback
        traceback.print_exc()
        raise

if __name__ == "__main__":
    main()
```
